```python
# imports 
import scipy
import scipy.constants as constants
import math
import numpy as np
```

```python
# constants 
h = constants.Planck # Js
kB = constants.Boltzmann # J/K
I = 4.67e-48  # kgm^2
pi = constants.pi 
R = constants.gas_constant # J/(molK)

print("h =",h,"Js")
print("kB =",kB,"J/K")
print("Moment of Inertia =",I,"kgm^2")
print("pi =",pi,"[-]")
print("R =",R,"J/(molK)")
```

```python
# corresponding rotational energy-level, 
def EnergyLevel_rotational(j): 
    nom = j*(j+1)*h**2
    denom = 8*pi**2*I
    return nom/denom
print("E1 =",EnergyLevel_rotational(1))

Temp = 77
levels_even = 200
levels_odd = 201

# partition functions Qp(even) and Qo(odd), dimensionless
# common part 
def Q_common(j,E,T): 
    x = 2*j+1 
    ex = math.exp(-E/(kB*T))
    return x*ex

# even 
def Q_para(T):
    s = sum(Q_common(j,EnergyLevel_rotational(j),T) for j in range (0,levels_even,2))
    return s
  
# odd 
def Q_ortho(T):
    s = 3*sum(Q_common(j,EnergyLevel_rotational(j),T) for j in range (1,levels_odd,2))
    return s

def Q_equil(T): 
    result = Q_ortho(T)+Q_para(T)
    return result 
          
def xpH2(T): 
    nom = Q_para(T)
    denom = Q_para(T) + Q_ortho(T) 
    return nom/denom
  
print("Qp =",Q_para(Temp)) 
print("Qo =",Q_ortho(Temp))
print("Qe =",Q_equil(Temp))
print("x_pH2 =",xpH2(Temp))
```

```python
# rotational contrbution E [K] (reduced by R, multiply result with R for J/mol as Unit)
def E_rotational_para(T):
    s = sum(EnergyLevel_rotational(j)/kB * Q_common(j,EnergyLevel_rotational(j),T) for j in range (0,levels_even,2))
    result = 1 / Q_para(T) * s 
    return result

def E_rotational_ortho(T):
    s = 3*sum(EnergyLevel_rotational(j)/kB * Q_common(j,EnergyLevel_rotational(j),T) for j in range (1,levels_odd,2))
    result = 1 / Q_ortho(T) * s
    return result

def E_rotational_equil(T): 
    s_even = sum(EnergyLevel_rotational(j)/kB * Q_common(j,EnergyLevel_rotational(j),T) for j in range (0,levels_even,2))
    s_odd = 3*sum(EnergyLevel_rotational(j)/kB * Q_common(j,EnergyLevel_rotational(j),T) for j in range (1,levels_odd,2))
    s = s_even + s_odd
    result = 1 / Q_equil(T) * s
    return result

def E_rotational_normal(T): 
    result = xpH2(T) * E_rotational_para(T) + (1-xpH2(T)) * E_rotational_ortho(T)
    return result

print("Rotational Energy (para): \n E/R =",E_rotational_para(Temp),"K")
print("Rotational Energy (ortho): \n E/R =",E_rotational_ortho(Temp),"K")
print("Rotational Energy (equil): \n E/R =",E_rotational_equil(Temp),"K")
print("Rotational Energy (normal): \n E/R =",E_rotational_normal(Temp),"K")

# internal energy U
# reference point set 0 at for all for now, change later 
U_ref_para = 0 
U_ref_ortho = 0 
U_ref_equil = 0 
U_ref_normal = 0 

U_ref_para = 0 
U_ref_ortho = 0 
U_ref_equil = 0

U_ref_para = 0 
U_ref_para = 0 
U_ref_para = 0 

U_ref_para = 0 
U_ref_para = 0 
U_ref_para = 0 

```
